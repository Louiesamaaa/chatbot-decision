```python
import sys
sys.path.append("..")
from src.env import Handoffenv, build_scenarios, is_bot_solvable
from src.classifier import IntentClassifier, load_abcd_data
from gymnasium.utils.env_checker import check_env
```

```python
data = load_abcd_data("../data/abcd_v1.1.json.gz")
clf = IntentClassifier.load("../models/intent_classifier.joblib")

scenarios = build_scenarios(data["train"], clf)
env= Handoffenv(scenarios, clf.model.classes_, seed = 0)


solvable_clf = IntentClassifier().fit(data["train"], label_fn=is_bot_solvable)

for k in [1, 2, 3]:
    print(k, "messages:", solvable_clf.evaluate(data["dev"], k, label_fn=is_bot_solvable))

print(solvable_clf.model.classes_)
print(solvable_clf.model.predict_proba(["what is your return policy?", "I forgot my password"]))

solvable_clf.save("../models/solvable_classifier.joblib")
```

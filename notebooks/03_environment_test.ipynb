```python
import sys
sys.path.append("..")

from src.classifier import IntentClassifier, load_abcd_data
from src.env import HandoffEnv, build_scenarios, CLARIFY
from gymnasium.utils.env_checker import check_env
```

```python
data = load_abcd_data("../data/abcd_v1.1.json.gz")
clf = IntentClassifier.load("../models/intent_classifier.joblib")

scenarios = build_scenarios(data["train"], clf)
env= HandoffEnv(scenarios, clf.model.classes_, seed = 0)

check_env(env)
print("Checked Environment Done!")
```

```python
obs, _ = env.reset(seed = 3)
env.render()
obs, reward, terminated, truncated, info = env.step(CLARIFY)
env.render()
print("reward:", reward)
```

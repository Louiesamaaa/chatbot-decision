```python
import sys
sys.path.append("..")

from src.classifier import IntentClassifier, load_abcd_data


data = load_abcd_data("../data/abcd_v1.1.json.gz")

clf = IntentClassifier()
clf.fit(data["train"])

for k in [1,2,3,4]:
    print(k, "messages:", clf.evaluate(data["dev"], k=k))

for msg in ["I want to cancel my order", "I want to return my order", "I want to change my shipping address","hi, I have a question"]:
    print(msg, "->", clf.predict(msg))

clf.save("../models/intent_classifier.joblib")
```
